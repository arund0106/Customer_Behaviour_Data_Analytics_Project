```python
# create a column called age_group

agegroup=['Young-Adult' , 'Adult' , 'Mid-Aged' ,'Senior']
df['age_group']=pd.qcut(df['age'] , q=4 , labels=agegroup)
```

```python
df[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
